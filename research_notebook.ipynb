```python
# Setup (Hidden - Dependencies & Configuration)

import os
import sys
from dotenv import load_dotenv
from IPython.display import display, HTML, Markdown
import warnings
warnings.filterwarnings('ignore')

# Add project to path
sys.path.insert(0, '/Users/ehsanchah/dev/multi_agent_sentiment_analysis-')

# Load environment variables
load_dotenv()

# Import agents
from src.agents.research_agent import ResearchAgent
from src.agents.auditor_agent import AuditorAgent
from src.agents.report_agent import ReportAgent
from crewai import Crew, Task

# Suppress verbose output during initialization
print("Environment initialized")
```

## INPUT: Research Parameters

Enter your research topic and target website below:

```python
# USER INPUT - Modify these values

RESEARCH_TOPIC = "byour topic"
TARGET_WEBSITE = "the website you want to target"

print(f"Research Topic: {RESEARCH_TOPIC}")
print(f"Target Website: {TARGET_WEBSITE}")
```

## PROCESSING (Hidden - Agent Initialization & Task Setup)

The research agents are now being initialized...

```python
# Initialize agents (hidden setup)
from src.tasks import create_research_tasks

research_agent = ResearchAgent()
auditor_agent = AuditorAgent()
report_agent = ReportAgent()

# Import and create tasks
search_task, audit_task, report_task = create_research_tasks(research_agent, auditor_agent, report_agent)

print("Agents initialized and tasks defined")
```

## RUN RESEARCH

Execute the multi-agent crew to search for articles and generate the report:

```python
# Run the crew
print("Starting research crew...\n")
print("=" * 60)

crew = Crew(
    agents=[research_agent, auditor_agent, report_agent],
    tasks=[search_task, audit_task, report_task],
    verbose=False  # Suppress verbose output for cleaner notebook display
)

result = crew.kickoff(inputs={"topic": RESEARCH_TOPIC, "website": TARGET_WEBSITE})

print("=" * 60)
print("\nResearch complete!")
```
